Part 3: Data Cleaning and Keyword Analysis

```python
#Create a new column for cleaned comments
TT["clean_text"] = TT["comment_text"]

#Remove website links from comments
TT["clean_text"] = TT["clean_text"].str.replace(r"http\S+","",regex=True)

#Display the original and cleaned comments for comparison
TT[["comment_text","clean_text"]].head()
```

An emoji appears in one of the comments in the table above. The following coding sequence is to clean the data and remove the unwanted emoji.

```python
#Install the emoji library
!pip install emoji
```

```python
#Import the emoji library
import emoji

#Remove emojis from the cleaned comments
TT["clean_text"] = TT["clean_text"].apply(emoji.replace_emoji)
```

```python
#Display the original and cleaned comments
TT[["comment_text","clean_text"]].head()
```

```python
#Remove punctuation from the comments
TT["clean_text"] = TT["clean_text"].str.replace(r"[^\w\s]","",regex=True)

#Display the first 5 rows
TT[["comment_text","clean_text"]].head()
```

The next step is converting all the comments to lowercase to allow for Python to recognise words easily when conducting keyword analysis. For example, LOVE, Love and love will be recognised as the same word when all is converted to lowercase.

```python
#Convert all comments to lowercase
TT["clean_text"] = TT["clean_text"].str.lower()

#Display the first 5 rows to check
TT[["comment_text","clean_text"]].head()
```

Applying text preprocessing using nltk.

```python
#Import NLTK tool for separating words
from nltk.tokenize import wordpunct_tokenize

#Split each cleaned comments into individual words
TT["tokens"] = TT["clean_text"].apply(wordpunct_tokenize)

#Display first 5 rows to check
TT[["clean_text","tokens"]].head()
```

The next sequence will be to remove stopwords. Stopwords are common words that do not tell us much about the main topic of the sentence. For example, "the", "is" and "and".

```python
#Import the list of common English stopwords
import nltk
from nltk.corpus import stopwords

#Download the English stopwords
nltk.download("stopwords")

#Store downloads in a list
stop_words = stopwords.words("english")
```

```python
#Create a function to remove stopwords
def remove_stopwords(words):

  #Create an empty list for the remaining words
  filtered = []

  #Go through each word
  for word in words:

    #Keep the word if it is not a stopword
    if word not in stop_words:
      filtered.append(word)

  #Return the remaining words
  return filtered

#Remove stopwords from all tokenised comments
TT["filtered_tokens"] = TT["tokens"].apply(remove_stopwords)

#Display first 5 rows to check
TT[["tokens","filtered_tokens"]].head()
```

The next sequence is lemmatization. Lemmatization refers to changing a word into its base form. For example, "cats" to "cat" and "running" to "run".

```python
#Import the tool for lemmatization
from nltk.stem import WordNetLemmatizer

#Download the WordNet dictionary
nltk.download("wordnet")

#Create a lemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
#Create a function to lemmatize words
def lemmatize_words(words):

  #Create an empty list for the results
  results = []

  #Go through each word
  for word in words:

    #Change the word to its base form
    results.append(lemmatizer.lemmatize(word))

  #Return the updated list
  return results

#Apply lemmatization to all the filtered comments
TT["lemmatized_tokens"] = TT["filtered_tokens"].apply(lemmatize_words)

#Display the first 5 rows to check
TT[["filtered_tokens","lemmatized_tokens"]].head()
```

The next sequence is to generate a word frequency table using Counter.

```python
#Import Counter to count words
from collections import Counter
```

```python
#Create an empty list to store all the words
all_words = []

#Go through each comment and collect the words
for comment in TT["lemmatized_tokens"]:
  all_words.extend(comment)

#Count how many times each word appears
word_count = Counter(all_words)

#Display the 10 most common words
word_count.most_common(10)
```

```python
#Create a table showing the 10 most common words
frequency_table = pd.DataFrame(word_count.most_common(10),columns=["Word","Frequency"])

#Display the frequency table
frequency_table
```

The next sequence will generate a word cloud to help visually identify words that appear the most often across the dataset.

```python
#Import the libraries for creating and displaying a word cloud
from wordcloud import WordCloud
import matplotlib.pyplot as plt

#Create a word cloud using the word frequency results
cloud = WordCloud(background_color="White").generate_from_frequencies(word_count)

plt.imshow(cloud)
plt.axis("off")
plt.show()
```

The word cloud shows that the most frequently used words include "trading", "year", "good", "invest" and "stock". This suggests that many of the TikTok comments are related to investing, trading and financial matters. Words such as "money", "profit" and "market" also appear frequently, suggesting that financial returns and market conditions may be common discussion topics. However, the word cloud only shows how frequently words appear and does not explain whether users are expressing positive or negative opinions. This will be further analysed in Part 4, where sentiment analysis is being done.

The next sequence will be calculating Term Frequency-Inverse Document Frequency (TF-IDF). This step is to help identify which words are particularly important, rather than simply counting how often they appear.

```python
#Import the TF-IDF tool
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
#Combine the cleaned words back into sentences
processed_comments = TT["lemmatized_tokens"].apply(" ".join)
```

```python
#Create the TF-IDF tool
tfidf = TfidfVectorizer()
```

```python
#Calculate the TF-IDF scores
tfidf_scores = tfidf.fit_transform(processed_comments)
```

```python
#Display how many comments were analysed
print("Total comments analysed:",tfidf_scores.shape[0])
```

```python
#Get all the words identified by TF-IDF
words = tfidf.get_feature_names_out()

#Calculate the average TF-IDF score for each word
scores = tfidf_scores.mean(axis=0).tolist()[0]
```

```python
#Create a table showing the words and their scores
tfidf_table = pd.DataFrame({"Word": words, "Score": scores})
```

```python
#Display the 10 words with the highest scores
tfidf_table.sort_values("Score", ascending=False).head(10)
```

Based on both the word cloud and TF-IDF analysis, words such as "trading", "investing" and "good" appear prominently in the TikTok comments. This suggests that many discussions revolve around financial investments and trading activities. The word "good" may indicate that users are sharing their opinions, although further sentiment analysis needs to be done to understand whether these opinions are positive or negative. Overall, the findings suggest that investing and trading are common discussion topics in the dataset.

Part 4: Sentiment and Topic Modeling

```python
#Import TextBlob to analyse sentiment
from textblob import TextBlob
```

```python
#Creating a sample comment
comment = "I love investing!"

#Analyse sentiment of the comment
result = TextBlob(comment).sentiment

#Display the results
print(result)
```

Polarity refers to whether a comment is positive, negative or neutral. -1 indicates negative, 0 indicates neutral and +1 indicates positive. The score above is 0.625 which leans more towards positive.

Subjectivity refers to whether a comment sounds more like a personal opinion or a factual statement. 0 indicates that the comment is more objective or fact based. 0.5 indicates that the comment is a mix of factual and opinion-based langague, whereas 1 indicates that the comment is more subjective or opionion based. In this case, the above analysis leans more towards the comment being somewhat objective.

```python
#Create a function to calculate polarity
def get_polarity(text):
  return TextBlob(text).sentiment.polarity
```

```python
#Create a function to calculate subjectivity
def get_subjectivity(text):
  return TextBlob(text).sentiment.subjectivity
```

```python
#Calculate subjectivity for all the comments
TT["subjectivity"] = TT["comment_text"].apply(get_subjectivity)

#Calculate polarity for all the comments
TT["polarity"] = TT["comment_text"].apply(get_polarity)
```

```python
#Display the first 5 rows to check the results
TT[["comment_text","polarity","subjectivity"]].head()
```

An interesting find is observed looking at index 1's comment "Too much stress with stocks bro". It received a polarity score of 0.2, suggesting a slightly positive sentiment. However, this comment appears to express frustration rather than positivity. This highlights a limitation of TextBlob, as it may not accurately understand the context, slang or emotions behind these commments.

```python
#Create a function to categorise sentiment
def get_sentiment(score):

  #Check if polarity is positive
  if score > 0:
    return "Positive"

  #Check if polarity is negative
  elif score < 0:
    return "Negative"

  #If score is 0, classify as neutral
  else:
    return "Neutral"
```

```python
#Apply functions to all comments
TT["sentiment"] = TT["polarity"].apply(get_sentiment)
```

```python
#Import the library for creating charts
import matplotlib.pyplot as plt
```

```python
#Calculate the percentage of each sentiment
sentiment_percent = TT["sentiment"].value_counts(normalize=True)*100
```

```python
#Arrange the sentiment categories
sentiment_percent = sentiment_percent.reindex(["Positive","Neutral","Negative"],fill_value=0)
```

```python
#Create a bar chart
sentiment_percent.plot(kind="bar")

#Add title and labels
plt.title("TikTok Comment Sentiment Analysis")
plt.xlabel("Sentiment")
plt.ylabel("Percentage of Comments(%)")

#Keep the categories labels horizontal
plt.xticks(rotation=0)

#Display bar chart
plt.show()
```

```python
#Create a histogram showing the distribution of polarity scores
plt.hist(TT["polarity"],bins=20,edgecolor="black")

#Add a title and labels to the histogram
plt.title("Distribution of TikTok Comment Polarity Scores")
plt.xlabel("Polarity Score")
plt.ylabel("Number of Comments")

#Display histogram
plt.show()
```

Based on the sentiment analysis, approximately 55% of the comments were classified as neutral, followed by 36% positive and 9% negative. The histogram also shows that most polarity scores are concentrated around 0, suggesting that strongly positive or negative comments are less common. Overall, the results suggest that TextBlob detected more positive than negative comments in the discussion surrounding investing and trading. However, a caveat that these findings should be interpreted carefully, as TextBlob may not accurately capture the intended meaning of comments containing slang, sarcasm or informal language. This is especially with reference to the above finding pertaining to the comment, "Too much stress with stocks bro" which was inaccurately depicted as a comment with positive sentiment.

```python
#Import Gensim to prepare words for topic modeling
from gensim import corpora
```

```python
#Create a dictionary using the lemmatized words
dictionary = corpora.Dictionary(TT["lemmatized_tokens"])
```

```python
#Display the number of unique words
print(("Number of unique words:"),len(dictionary))
```

The next sequence will require creating a corpus record which words appear in each comment and how frequently they appear.

```python
#Create a corpus from the cleaned comments
corpus = []

#Convert each comment into word IDs and frequencies
for comment in TT["lemmatized_tokens"]:
  corpus.append(dictionary.doc2bow(comment))
```

```python
#Import LDA model from Gensim
from gensim.models import LdaModel

#Create an LDA Model with 3 topics
lda_model = LdaModel(corpus=corpus,id2word=dictionary,num_topics=3,random_state=42,passes=10)
```

```python
#Display the keywords for each topic
lda_model.print_topics()
```

Creating another LDA with 5 topics for a more thorough comparison.

```python
#Creating another LDA model with 5 topics
lda_model_5 = LdaModel(corpus=corpus,id2word=dictionary,num_topics=5,random_state=42,passes=10)
```

```python
#Display the keywords for all 5 topics
lda_model_5.print_topics()
```

Experimented with the 3-topic and 5-topic LDA models to identify common discussion themes in the comments. Both models showed that investing and trading were recurring subjects. However, the 5-topic model appeared to provide more specific themes. Therefore, selecting the 5-topic model for further interpretation, although some keywords still overlap across the topics.

```python
#Display 5 top words for each topic
lda_model_5.print_topics(num_words=5)
```
